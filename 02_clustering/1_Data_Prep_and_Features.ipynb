```python
# ==============================================================================
# 1. DATA PREP & FORENSIC FEATURES
# ==============================================================================
import pandas as pd
import numpy as np
import os
import warnings

# Setup
warnings.filterwarnings('ignore')

# RUTAS:

BASE_DIR = "../" 

# Rutas de Archivos (Apuntando a NeoModelos)
INPUT_CSV = "../../02_Clustering/outputs/minas_con_tiempos_puertos.csv"
OUTPUT_FILE = "../../02_Clustering/outputs/1_clustering_input_ready.csv"

print(f"📂 Buscando archivo en: {os.path.abspath(INPUT_CSV)}")

if os.path.exists(INPUT_CSV):
    df_raw = pd.read_csv(INPUT_CSV)
    print(f">> Datos Cargados: {len(df_raw)} filas.")
else:
    print(">> ERROR: No se encuentra el archivo.")
    print(f"   Directorio actual: {os.getcwd()}")
    print(f"   Ruta buscada: {INPUT_CSV}")
    raise FileNotFoundError(f"Input file not found: {INPUT_CSV}")

# --- 1. FILTROS Y LIMPIEZA ---
print(">> Aplicando filtros...")

# Solo minas ACTIVAS y recursos específicos (Cobre/Litio)
df_all = df_raw[df_raw['Estado'] == 'ACTIVA'].copy()
df_all = df_all[(df_all['RecursoPrimarioInstalacion'] == 'COBRE') | 
                (df_all['RecursoMineroInstalacion'] == 'SALMUERA (LITIO)')]

# Eliminar filas sin geo-data crítica
df_all = df_all.dropna(subset=['IdFaena', 'Latitud', 'Longitud', 'RegionFaena'])

# Estandarizar IDs
df_all['IdFaena'] = df_all['IdFaena'].astype(int).astype(str)

# Crear bandera estratégica (Categoría A)
df_all['Es_Estrategica'] = df_all['CategoriaFaena'] == 'CATEGORIA A'

# Normalizar Nombres de Región (Crítico para que coincida con las llaves del NPZ después)
df_all['Region_Norm'] = df_all['RegionFaena'].str.upper()\
                                             .str.replace('REGIÓN DE ', '')\
                                             .str.replace('REGIÓN DEL ', '')\
                                             .str.strip()

print(f"📊 Faenas Activas después del filtro: {len(df_all)}")

# --- 2. EXPORTAR PARA EL MOTOR DE CLUSTERING ---
df_all.to_csv(OUTPUT_FILE, index=False)
print(f">> Archivo preparado guardado en: {OUTPUT_FILE}")
```
